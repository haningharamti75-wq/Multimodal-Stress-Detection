```python
def print_metrics(title, df):
    print("=" * 60)
    print(title)
    print("=" * 60)

    if "accuracy" in df.columns:
        print(f"Accuracy           : {df['accuracy'].mean():.4f}")
    elif "acc" in df.columns:
        print(f"Accuracy           : {df['acc'].mean():.4f}")

    if "balanced_acc" in df.columns:
        print(f"Balanced Accuracy  : {df['balanced_acc'].mean():.4f}")
    elif "bal_acc" in df.columns:
        print(f"Balanced Accuracy  : {df['bal_acc'].mean():.4f}")

    if "precision" in df.columns:
        print(f"Precision          : {df['precision'].mean():.4f}")

    if "recall" in df.columns:
        print(f"Recall             : {df['recall'].mean():.4f}")

    print(f"F1-score           : {df['f1'].mean():.4f}")
    print()


print_metrics("CNN", cnn_metrics)
print_metrics("ST-GCN", stgcn_metrics)
print_metrics("BiLSTM", bilstm_metrics)
print_metrics("Two-Stream Fusion", two_stream_results)
print_metrics("Weighted Late Fusion", late_fusion_results)
```

```python
# Create a dictionary: participant -> threshold
cnn_thresholds = dict(
    zip(
        cnn_metrics["participant"],
        cnn_metrics["threshold"]
    )
)

# Add the threshold to every prediction row
cnn_probs["threshold"] = cnn_probs["pid"].map(cnn_thresholds)

print(cnn_probs.head())

print("\nMissing thresholds:",
      cnn_probs["threshold"].isna().sum())
```

```python
cnn_probs["prediction"] = (
    cnn_probs["prob_stress"] >= cnn_probs["threshold"]
).astype(int)

cnn_probs.head()
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

rows = []

for pid, df in cnn_probs.groupby("pid"):

    y_true = df["true_label"]
    y_pred = df["prediction"]

    rows.append({
        "participant": pid,
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_acc": balanced_accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "f1": f1_score(y_true, y_pred, zero_division=0),
    })

cnn_recalculated = pd.DataFrame(rows)

cnn_recalculated.head()
```

```python
comparison = cnn_recalculated.merge(
    cnn_metrics,
    on="participant",
    suffixes=("_calc", "_saved")
)

print("Maximum Accuracy difference:",
      (comparison["accuracy_calc"] - comparison["accuracy_saved"]).abs().max())

print("Maximum Balanced Accuracy difference:",
      (comparison["balanced_acc_calc"] - comparison["balanced_acc_saved"]).abs().max())

print("Maximum Precision difference:",
      (comparison["precision_calc"] - comparison["precision_saved"]).abs().max())

print("Maximum Recall difference:",
      (comparison["recall_calc"] - comparison["recall_saved"]).abs().max())

print("Maximum F1 difference:",
      (comparison["f1_calc"] - comparison["f1_saved"]).abs().max())
```
